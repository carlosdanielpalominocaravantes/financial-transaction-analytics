```python
import pandas as pd
import numpy as np

print("🚀 Iniciando Pipeline de la Cultura Zero-Manual...")

# 1. SIMULACIÓN DE DATOS CRUDOS (Generamos 10,000 filas de prueba para el script)
np.random.seed(42)
n_rows = 10000

data = {
    'transaction_id': [f"TX-{100000 + i}" for i in range(n_rows)],
    'date': pd.date_range(start='2026-01-01', periods=n_rows, freq='min'),
    'account_id': np.random.randint(5000, 5050, size=n_rows),
    'category': np.random.choice(['Ingreso', 'Gasto', 'Devolución', 'Inversión', None], size=n_rows, p=[0.4, 0.4, 0.1, 0.05, 0.05]),
    'amount': np.random.uniform(-100, 5000, size=n_rows),
    'status': np.random.choice(['Completed', 'Pending', 'Failed'], size=n_rows, p=[0.85, 0.10, 0.05])
}

df_raw = pd.DataFrame(data)

# Introducir duplicados y anomalías contables a propósito para limpiarlos
df_raw = pd.concat([df_raw, df_raw.iloc[:150]], ignore_index=True) # 150 filas duplicadas
df_raw.loc[df_raw['category'] == 'Gasto', 'amount'] = df_raw.loc[df_raw['category'] == 'Gasto', 'amount'] * -1 # Asegurar montos negativos para gastos

print(f"📊 Dataset crudo creado con {df_raw.shape[0]} registros financieros.")

# ==========================================
# 2. PROCESO DE LIMPIEZA Y TRANSFORMACIÓN CONTABLE
# ==========================================

# Identificar y eliminar transacciones duplicadas
duplicados_detectados = df_raw.duplicated(subset=['transaction_id']).sum()
df_clean = df_raw.drop_duplicates(subset=['transaction_id'], keep='first').copy()
print(f"🧹 Transacciones duplicadas eliminadas: {duplicados_detectados}")

# Manejo de valores faltantes en categorías contables
nulos_categoria = df_clean['category'].isnull().sum()
df_clean['category'] = df_clean['category'].fillna('No Clasificado')
print(f"🧩 Categorías contables faltantes rellenadas como 'No Clasificado': {nulos_categoria}")

# Detección de valores atípicos (Outliers) en montos usando desviaciones estándar
mean_amount = df_clean['amount'].mean()
std_amount = df_clean['amount'].std()
outliers = df_clean[(df_clean['amount'] > (mean_amount + 3 * std_amount)) | (df_clean['amount'] < (mean_amount - 3 * std_amount))]
print(f"⚠️ Alerta de Auditoría: Se detectaron {len(outliers)} transacciones con montos atípicos (Fuera de 3 desviaciones estándar).")

# Guardar dataset procesado y limpio para el Día 2
df_clean.to_csv('clean_financial_transactions.csv', index=False)
print("💾 Archivo 'clean_financial_transactions.csv' exportado con éxito y listo para la base de datos.")
```
